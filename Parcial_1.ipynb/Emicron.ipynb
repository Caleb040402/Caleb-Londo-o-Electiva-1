```python
import pandas as pd
import numpy as np
```

```python
df_Emicron = pd.read_csv("https://github.com/niconomist98/DataAnalyticsUQ/raw/refs/heads/main/Datos/EMICRON/emicron_identificacion.csv", encoding='latin-1', sep=',')
```

```python
df_Emicron
```

```python
#pregunta 1
#¿ cuanto micronegocios hay en el quindio cuyos propietarios son hombres entre los 18 y 30 años?
#Encontrar la cantidad de micronegocios en Colombia
total = df_Emicron["F_EXP"].sum()
print("Cantidad de micronegocios en Colombia:", int(total))
```

```python
#Buscar si estan las variables que necesitamos
df_Emicron.columns
```

```python
#Micronegocios en el Quindio cuyos propietarios son hombres entre 18 y 30 años
micronegocios_HQ = df_Emicron[(df_Emicron["COD_DEPTO"] == 63) & (df_Emicron["P35"] == 1) & (df_Emicron["P241"].between(18, 30))]
print("Cantidad de micronegocios en el Quindio cuyos propietarios son hombres entre 18 y 30 años:", int(micronegocios_HQ["F_EXP"].sum()))
```

```python
#2 Cuantos micronegocios de la industria manufacturera existen en el departamento de risaralda cuyos propietarias son mujeres entre los 20 y 30 años?
q2= df_Emicron[(df_Emicron["COD_DEPTO"] == 66) & (df_Emicron["P35"] == 2) & (df_Emicron["P241"].between(20, 30)) & (df_Emicron["GRUPOS4"] == 2)]

total_micronegocios_Manufacturera_RM = q2["F_EXP"].sum()
print("Cantidad de micronegocios de la industria manufacturera en el departamento de risaralda cuyos propietarias son mujeres entre los 20 y 30 años:", int(total_micronegocios_Manufacturera_RM))
```

```python
#3. Entre las ciudades de Armenia, Pereira y Manizales, ¿cuál tiene el mayor número de micronegocios en el sector de comercio a cuyas propetarias son mayores a 40 años?

#Micronegocios en Armenia 
#ARMENIA 
micronegocios_comercio_A40 = df_Emicron[(df_Emicron["AREA"] == 63) & (df_Emicron["P35"] == 2) & (df_Emicron["P241"] > 40) & (df_Emicron["GRUPOS4"] == 3)]
total_micronegocios_comercio_A40 = micronegocios_comercio_A40["F_EXP"].sum()
print(f"El total de micronegocios del sector comercio en la ciudad de Armenia, cuyas propietarias son mujeres mayores a 40 años es: {micronegocios_comercio_A40['F_EXP'].sum():.0f}")
```

```python
#Micronegocios en Pereira
micronegocios_comercio_P50 = df_Emicron[(df_Emicron["AREA"] == 66) & (df_Emicron["P35"] == 2) & (df_Emicron["P241"] > 40) & (df_Emicron["GRUPOS4"] == 3)]
total_micronegocios_comercio_P50 = micronegocios_comercio_P50["F_EXP"].sum()
print(f"El total de micronegocios del sector comercio en la ciudad de Pereira, cuyas propietarias son mujeres mayores a 40 años es: {micronegocios_comercio_P50['F_EXP'].sum():.0f}")
```

```python
#Micronegocios en Manizales
#MANIZALES
micronegocios_comercio_M50 = df_Emicron[(df_Emicron["AREA"] == 17) & (df_Emicron["P35"] == 2) & (df_Emicron["P241"] > 40) & (df_Emicron["GRUPOS4"] == 3)]
total_micronegocios_comercio_M50 = micronegocios_comercio_M50["F_EXP"].sum()
print(f"El total de micronegocios del sector comercio en la ciudad de Manizales, cuyas propietarias son mujeres mayores a 40 años es: {micronegocios_comercio_M50['F_EXP'].sum():.0f}")
```

```python
#4.  En la ciudad de armenia , cuanto micronegicos hay cuyos propietarios son hombres que no tienen nombre comercial
micronegocios_AXM_snc = df_Emicron[(df_Emicron["AREA"] == 63) & (df_Emicron["P35"] == 1) & (df_Emicron["P3035"] == 2)]
total_micronegocios_AXM_snc = micronegocios_AXM_snc["F_EXP"].sum()
print(f"El total de micronegocios en la ciudad de Armenia, cuyos propietarios son hombres que no tienen nombre comercial es: {micronegocios_AXM_snc['F_EXP'].sum():.0f}")
```
